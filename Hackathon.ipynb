roads

```python
data = pd.read_csv('/content/features_with_labels_fix.csv')

data = data.drop('filename',axis=1)

X = data.drop(['houses',	'roads',	'ponds',	'Electric Poles'],axis=1)
y = data['roads']
```

```python
from sklearn.ensemble import StackingClassifier,RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

estimators = [
    ('rf', RandomForestClassifier(n_estimators=80, random_state=42,max_depth=15,min_samples_split=2,min_samples_leaf=2)),
    ('xgb', XGBClassifier(random_state=42,n_estimators=80,reg_alpha=0.5,min_child_weight=10,reg_lambda=5))
]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=12, stratify=y
)

model = StackingClassifier(
    stack_method='predict_proba',estimators=estimators,final_estimator=LogisticRegression())

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print(classification_report(y_test,y_pred,digits=6))
```

```python
import joblib

joblib.dump(model,'xgb_model_roads_stack.pkl')
```

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold

sk = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
cv_score = cross_val_score(model, X, y, cv=sk)
print(cv_score.mean())
```

Electric Poles

```python
data = pd.read_csv('/content/features_with_labels_fix.csv')

data = data.drop('filename',axis=1)

X = data.drop(['houses',	'roads',	'ponds',	'Electric Poles'],axis=1)
y = data['Electric Poles']
```

```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=12, stratify=y
)

model = XGBClassifier(random_state=42,scale_pos_weight=4,reg_lambda=5,reg_alpha=0.5,
                      min_child_weight=10,n_estimators=200,learning_rate=0.7,max_depth=10)

model.fit(X_train,y_train)

y_pred = model.predict(X_test)

y_pred2 = (model.predict_proba(X_test)[:,1]>0.3).astype(int)

print(classification_report(y_test, y_pred2,digits=4))
```

```python
import joblib

joblib.dump(model, 'xgb_model_electric_poles.pkl')
```

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold

sk = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
cv_score = cross_val_score(model, X, y, cv=sk)
print(cv_score.mean())
```
